## 2. Load the Raw Dataset

The original UCI dataset uses a semicolon (`;`) as the column separator. Missing measurements are represented by `?`, so we convert those values to Pandas `NaN` values while loading the data.


```python
file_path = "../data/raw/household_power_consumption.txt"

df = pd.read_csv(
    file_path,
    sep=";",
    na_values="?",
    low_memory=False
)

df.head()
```

## 9. Missing Values

Count the missing values in every column. Missing values will be handled systematically in the data-cleaning notebook.


```python
missing_values = df.isnull().sum()
missing_values
```

## 10. Missing-Value Percentage

Calculate the percentage of missing observations in each column. This helps us understand the severity of the missing-data problem.


```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).sort_values(ascending=False)

missing_percentage
```
